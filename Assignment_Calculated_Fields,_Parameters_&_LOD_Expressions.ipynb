#Calculated Fields, Parameters & LOD Expressions


# Question 1: What is a Calculated Field and why is it important in business dashboards?

## Answer

A Calculated Field in Tableau is a custom field created using formulas and functions to perform calculations on existing data. It allows analysts to derive new insights without modifying the original dataset.

Calculated fields are important because they help businesses create meaningful metrics such as Profit Margin, Discount Percentage, Sales Growth, and Performance Categories.

### Importance in Business Dashboards

- Creates custom business metrics.
- Simplifies complex calculations.
- Improves decision-making.
- Reduces dependency on source data modifications.
- Enables dynamic and real-time analysis.

### Example

```text
Profit Margin = Profit / Sales
```

This calculation helps management understand profitability instead of looking only at sales figures.

## Conclusion

Calculated Fields transform raw data into actionable business insights and are essential for creating effective dashboards.


# Question 2: Why can’t parameters change visuals directly in Tableau?

## Answer

Parameters in Tableau are dynamic input controls that allow users to enter values or make selections. However, parameters do not directly interact with visualizations.

A parameter only stores a value. To affect a chart or dashboard, it must be connected to a Calculated Field, Filter, Reference Line, or other Tableau feature.

### Example

A parameter called "Top N Products" may contain the value 10.

The parameter alone will not change the chart.

A calculated field must use the parameter value to filter and display the top 10 products.

### Why Parameters Cannot Change Visuals Directly

- Parameters only store values.
- They are independent of the dataset.
- They require calculations or filters to impact visualizations.

## Conclusion

Parameters act as user inputs, while calculated fields and filters use those inputs to modify visualizations and analysis.


# Question 3: When should EXCLUDE LOD be used?

## Answer

EXCLUDE LOD (Level of Detail) is used when we want to remove a dimension from a calculation while keeping other dimensions in the view.

It helps compare detailed values against a broader aggregated value.

### Syntax

```text
{ EXCLUDE [Dimension] : Aggregation() }
```

### Example

If a report displays Sales by Category and Sub-Category, EXCLUDE can calculate sales at the Category level while ignoring Sub-Category.

### Business Use Case

A manager wants to compare each product's sales with the overall category average.

EXCLUDE allows Tableau to ignore the product detail and calculate the category-level value.

## Conclusion

EXCLUDE LOD is useful when detailed dimensions need to be removed from calculations to create higher-level comparisons.


# Question 4: Difference between FIXED and INCLUDE LOD Expressions

## Answer

Both FIXED and INCLUDE are Level of Detail expressions, but they work differently.

| Feature | FIXED LOD | INCLUDE LOD |
|----------|-----------|------------|
| Calculation Level | Fixed at specified dimension | Adds dimensions to current view |
| Affected by View Dimensions | No | Yes |
| Affected by Filters | Only Context Filters | Most Filters |
| Use Case | Independent benchmarks | Detailed calculations within summaries |

### FIXED Example

```text
{ FIXED [City] : AVG([Sales]) }
```

Calculates average sales for each city regardless of dimensions displayed.

### INCLUDE Example

```text
{ INCLUDE [Product_Name] : AVG([Sales]) }
```

Calculates average sales including product-level details even if Product Name is not displayed.

## Conclusion

FIXED creates stable benchmark calculations, while INCLUDE adds detail to calculations without requiring the dimension to be visible in the view.


# Task 1: Delivery Performance Classification (Calculated Field)

## Business Need

# Task 5: Customer/Product Detail Impact (INCLUDE LOD)

## Business Need

Analyze average sales at the product level while displaying results at higher aggregation levels.

## INCLUDE LOD Formula

```text
{ INCLUDE [Product_Name] : AVG([Sales]) }
```

## Purpose

This calculation considers product-level sales while allowing results to be viewed at Category or City level.

## Benefits

- Maintains product-level detail.
- Supports higher-level reporting.
- Provides more accurate averages.
- Improves product performance analysis.

## Example Use Case

Management wants to view Category performance while still considering the impact of individual products.

INCLUDE LOD ensures product-level information is included in the calculation even when Product Name is not visible.

## Conclusion

INCLUDE LOD helps organizations perform more detailed analysis without cluttering reports with excessive dimensions.
